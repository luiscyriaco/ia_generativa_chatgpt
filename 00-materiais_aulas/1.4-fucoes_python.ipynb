## Dominando Funções em Python: Da Base à Arquitetura de Sistemas

Até agora, escrevíamos todo o código de forma linear. Mas quando um sistema cresce, repetir o mesmo código várias vezes é inviável. É aí que entram as **Funções**.

Uma função funciona exatamente como uma **receita de bolo**:
1. **Entrada (Parâmetros):** Você passa os dados *(como separar os ingredientes na bancada antes de começar)*.

2. **Processamento:** A função executa um conjunto de regras *(como misturar e assar tudo na ordem certa)*.

3. **Saída (`return`):** Ela devolve o resultado pronto para o restante do código *(como tirar o bolo pronto do forno para ser servido na mesa)*.

Antes de criarmos nossa primeira função real, vamos entender o "esqueleto" de como uma função é construída em Python:

```Python
# Os parâmetros e o return é opcional na estrutura de uma função
def nome_da_funcao(parametro_1, parametro_2):
    # Bloco de código
    resultado = parametro_1 + parametro_2
    return resultado  # O que a função devolve para fora

#### Etapa 1: A função mais simples (Sem parâmetros e sem return)
É aquela que apenas executa uma tarefa fixa e mostra o resultado direto na tela (ou no terminal). Ela não recebe nada de fora e não devolve nada.

````python
# Definindo a função básica
def exibir_cabecalho():
  print("==================================")
  print("   SISTEMA DE GERENCIAMENTO V1.0  ")
  print("==================================")


# Chamando a função para ela executar o que sabe fazer
exibir_cabecalho()
````

#### Etapa 2: Função com Parâmetros (Recebendo dados de fora)
Aqui, a função ganha flexibilidade. Ela passa a aceitar dados (parâmetros) entre os parênteses para personalizar a tarefa que vai executar.

````python
# A função agora recebe um 'nome' como parâmetro para personalizar a saudação
def saudacao(nome):
  print(f"Olá, {nome}! Seja bem-vindo(a) ao sistema.")


# Chamando a função e passando valores diferentes
saudacao("Ana")
saudacao("Carlos")
````

#### Etapa 3: Função com return (Devolvendo um resultado)
Em vez de apenas mostrar o resultado na tela, a função processa uma informação e devolve (return) o valor para quem a chamou, permitindo guardá-lo em uma variável.

````python
# A função calcula a média e devolve o resultado com o 'return'
def calcular_media(nota1, nota2):
  soma = nota1 + nota2
  media = soma / 2
  return media


# Chamando a função e guardando o valor retornado em uma variável
resultado_final = calcular_media(8.0, 6.0)
print("A média do aluno é:", resultado_final)
````

#### Etapa 4: Função Completa (Parâmetros + Processamento + return)
Juntando tudo: a função recebe múltiplos dados, faz cálculos ou regras lógicas complexas e devolve o resultado final pronto para ser usado (exatamente o padrão que usamos no nosso gerador de senhas ou em códigos profissionais).

````python
# Função completa que verifica se um número é par
def e_par(numero):
  if numero % 2 == 0:
    return True
  else:
    return False


# Usando o retorno da função em uma estrutura de decisão
if e_par(4):
  print("O número é par!")
else:
  print("O número é ímpar!")
````
