## Dominando Listas em Python

No início, vimos que uma lista é uma "caixinha" que guarda vários itens em ordem, acessados por índices numéricos [`0, 1, 2...`]. Mas no dia a dia da programação, usamos listas para lidar com **lotes de dados** (como uma lista de arquivos em uma pasta ou uma lista de textos extraídos).

### O que você precisa dominar com listas:
1. **Adicionar dinamicamente:** Saber usar `.append()` para inserir novos dados dentro da lista em tempo de execução.
2. **Varredura com `for`:** Percorrer item por item para aplicar uma regra ou condição.
3. **Filtros com Loop:** Usar um `for` combinado com uma condição (`if`) para selecionar apenas os itens que atendem a um critério específico.

---
## Exercício Prático — Filtrando Arquivos Válidos

**O Contexto:** Imagine que você recebeu uma lista de arquivos variados em um diretório do computador, mas precisa separar apenas os arquivos `.pdf` para processamento posterior. Vamos criar um script para fazer esse filtro de forma automatizada.

```python
# Lista de arquivos encontrados no diretório
arquivos_enviados = ["manual.pdf", "foto.png", "contrato.PDF", "virus.exe", "relatorio.pdf", "anotacoes.txt"]

# 1. Crie uma lista vazia chamada 'pdfs_validos'
pdfs_validos = []

# 2. Percorra cada arquivo da lista 'arquivos_enviados' usando um loop for
for arquivo in arquivos_enviados:
    
    # 3. Use uma condição para verificar se o nome do arquivo termina com '.pdf' (Dica: use .lower() para evitar erro com maiúsculas/minúsculas)
    if arquivo.lower().endswith(".pdf"):
        
        # 4. Se for PDF, adicione o arquivo na lista 'pdfs_validos' usando .append()
        pdfs_validos.append(arquivo)

# Exibe o resultado final para ver a execução
print("Arquivos brutos:", arquivos_enviados)
print("Apenas PDFs filtrados:", pdfs_validos)
```

---
### O que esse exercício fixa na prática:
* O uso do `for` iterando sobre elementos reais de uma lista.
* Métodos de string essenciais como `.lower()` e `.endswith()`.
* O método `.append()` populando uma estrutura de dados do zero.

# Dominando Dicionários e o Método .get()

Até agora vimos listas, onde guardamos os itens usando números de posição (índices). Mas e quando precisamos guardar dados rotulados, como o cadastro de um usuário, as configurações de um sistema ou informações detalhadas? 

Para isso, usamos os **Dicionários**. Eles funcionam com um par de **Chave** (o rótulo, ex: `"nome"`) e **Valor** (ex: `"Ana"`). Em Python, usamos chaves `{}` para criá-los.

```python
# Lista contendo o perfil de dois usuários diferentes
usuarios = [
    {
        "nome": "Carlos Silva",
        "idade": 28,
        "ativo": True,
        "cargo": "Desenvolvedor"
    },
    {
        "nome": "Ana Souza",
        "idade": 32,
        "ativo": True,
        "cargo": "Engenheira de Dados"
    }
]

# Acessando os dados de cada usuário na lista de dicionários
print("Primeiro Usuário:", usuarios[0]["nome"], "| Cargo:", usuarios[0]["cargo"])
print("Segundo Usuário:", usuarios[1]["nome"], "| Cargo:", usuarios[1]["cargo"])
```

## O Perigo do `KeyError` e a Salvação com `.get()`

Quando trabalhamos com dados que vêm de arquivos externos ou configurações, nem sempre temos certeza se todas as chaves esperadas foram preenchidas. 

Se tentarmos acessar uma chave que não existe da forma tradicional (`dicionario["chave_inexistente"]`), o Python vai interromper o programa com um erro chamado `KeyError`.

Para evitar que o sistema quebre, usamos o método **`.get()`**. Ele tenta buscar a chave com segurança e, se ela não existir, retorna um valor padrão (como um texto alternativo ou vazio) em vez de dar erro.

```python
# Simulando um cadastro que veio incompleto (faltando a chave 'telefone')
cadastro_incompleto = {
    "nome": "Mariana",
    "email": "mariana@email.com"
    # note que a chave 'telefone' não existe aqui!
}

# JEITO PERIGOSO (Se a chave não existir, o programa crasha):
# print(cadastro_incompleto["telefone"]) 

# JEITO SEGURO (Usando o .get()):
# Se 'telefone' não existir, ele retorna "Não informado" em vez de quebrar
telefone = cadastro_incompleto.get("telefone", "Não informado")

print("Telefone extraído com segurança:", telefone)
```

Exercício Prático — Processando Múltiplos Cadastros

Imagine que você está desenvolvendo um sistema que recebeu uma lista com vários registros de formulários. Cada registro possui um ID e um perfil de usuário, mas algumas informações (como o telefone) podem ser opcionais ou faltar em alguns cadastros.

Complete o código abaixo usando um loop for e o método .get() para extrair o id, o nome e o telefone de forma segura, evitando erros caso algum campo não tenha sido preenchido.

```python
# Lista de registros recebidos (contendo múltiplos IDs: 1042 e 1043)
cadastros = [
    {
        "id": 1042,
        "perfil": {
            "nome": "Beatriz Souza",
            "email": "beatriz@email.com"
        }
    },
    {
        "id": 1043,
        "perfil": {
            "nome": "Carlos Silva",
            "email": "carlos@email.com"
            # note que o Carlos não tem telefone cadastrado
        }
    }
]

# Percorrendo a lista de cadastros com um loop for
for item in cadastros:
    # 1. Extrai o ID de forma segura
    id_formulario = item.get("id", 0)
    
    # 2. Extrai o dicionário 'perfil' com segurança (retorna {} se não existir)
    perfil_extraido = item.get("perfil", {})
    
    # 3. Pega o nome do usuário dentro do perfil
    nome_usuario = perfil_extraido.get("nome", "Visitante")
    
    # 4. Tenta pegar o telefone (que pode não existir para alguns IDs)
    telefone_usuario = perfil_extraido.get("telefone", "Não cadastrado")
    
    # Exibe o resultado processado de cada ID
    print(f"ID: {id_formulario} | Usuário: {nome_usuario} | Telefone: {telefone_usuario}")
```
