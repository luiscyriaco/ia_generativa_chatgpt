Até agora, nossos códigos rodavam perfeitamente quando tudo saía como o esperado. Mas e no mundo real, quando o usuário digita uma letra em vez de um número, ou tentamos abrir um arquivo que não existe na pasta?

Sem proteção, o programa inteiro **quebra (dá crash)** e encerra a execução de forma abrupta. 

O `try` e o `except` funcionam como um **paraquedas** ou uma rede de segurança:
1. **`try` (Tente):** O Python tenta executar um bloco de código que é arriscado.
2. **`except` (Caso dê erro):** Se algo der errado no `try`, o Python não quebra o programa; ele desvia o fluxo para cá e executa um plano de salvamento.

```python
# Exemplo clássico: Tentando converter o texto digitado pelo usuário em número inteiro
entrada_usuario = "abc"

try:
    # O Python tenta fazer a conversão arriscada
    numero_convertido = int(entrada_usuario)
    print("Conversão feita com sucesso:", numero_convertido)
    
except:
    # Se der erro (porque "abc" não é número), ele cai aqui em vez de quebrar
    print("Erro: Você precisa digitar apenas números válidos!")
```

## Exercício Prático — Blindando o Servidor contra Divisão por Zero

Imagine que você está criando uma calculadora para um painel web e o sistema recebe um divisor enviado por um cliente. Se o cliente enviar o número `0`, a matemática quebra e o sistema fecha com erro.

Complete o código abaixo usando `try` e `except` para capturar qualquer erro de divisão e retornar uma mensagem amigável para o usuário.

```python
# Função que realiza uma divisão de forma segura
def dividir_seguro(x, y):
    try:
        # 1. Tente realizar a divisão matemática
        resultado = x / y
        return f"Resultado da divisão: {resultado}"
        
    except:
        # 2. Se o Python encontrar um erro (como dividir por zero), retorne uma mensagem de aviso amigável
        return "Erro: Não é possível dividir um número por zero!"

# Testando com valores normais e valores problemáticos
print(dividir_seguro(10, 2))  # Deve funcionar normalmente
print(dividir_seguro(10, 0))  # Vai tentar dividir por zero, mas o except vai salvar!
```
