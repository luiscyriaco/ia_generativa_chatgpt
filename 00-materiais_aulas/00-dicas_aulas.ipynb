## O Ciclo Natural de Aprendizagem Aplicado ao Python

### 1. A Fala: Execução Imediata (Input / Output)
* **Conceito:** O aluno faz o código rodar e vê o resultado na tela imediatamente, sem se preocupar com regras profundas.
* **Prática:** Uso do `print()`, `input()` e scripts rápidos no terminal para gerar feedback visual instantâneo.

---

### 2. A Estrutura: Sintaxe e Blocos Lógicos (A Gramática)
* **Conceito:** O estudo formal das regras de organização e dos padrões que formam a linguagem.
* **Prática:** Variáveis, tipos de dados, condicionais (`if/else`), loops (`for/while`) e criação de funções (`def`).

---

### 3. A Semântica: Lógica e Modelagem (O Significado)
* **Conceito:** O entendimento profundo de como e por que o código resolve o problema, focando em arquitetura e lógica.
* **Prática:** Tratamento de erros (`try/except`), organização de código limpo (*Clean Code*) e tradução de problemas reais em algoritmos.
