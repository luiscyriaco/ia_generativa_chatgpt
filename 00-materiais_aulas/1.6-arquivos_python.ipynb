## Persistência de Dados: Salvando e Lendo Arquivos em Python

Até agora, todas as informações que manipulávamos (listas, dicionários) morriam assim que o código terminava de rodar, pois ficavam apenas na memória RAM do computador. 

Para criar sistemas reais, precisamos salvar esses dados de forma permanente no disco (em arquivos). Vamos seguir a evolução natural do armazenamento: do texto simples (`.txt`) até os formatos estruturados que usamos em sistemas e APIs (`.csv` e `.json`).

### 1. O Formato Mais Simples: Arquivos de Texto (`.txt`)

O comando nativo do Python para manipular arquivos é o `open()`. Ele precisa de dois argumentos principais:
1. O **nome do arquivo** que você quer criar ou abrir.
2. O **modo de operação**:
   * `"w"` (Write/Escrita): Cria um arquivo novo ou **sobrescreve** o antigo se ele já existir.
   * `"r"` (Read/Leitura): Lê o conteúdo de um arquivo existente.
   * `"a"` (Append/Adicionar): Adiciona novas linhas no final de um arquivo existente sem apagar o que já está lá.

Sempre usamos a estrutura `with open(...) as arquivo:` porque ela fecha o arquivo automaticamente para nós após o uso, evitando travamentos de sistema.

```python
# --- ESCREVENDO EM UM ARQUIVO .TXT ---
# O modo 'w' cria o arquivo 'relatorio.txt' na sua pasta
with open("relatorio.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("Primeira linha: Sistema iniciado com sucesso.\n")
    arquivo.write("Segunda linha: Dados salvos pelo Python.\n")

print("Arquivo .txt criado e escrito com sucesso!")


# --- LENDO UM ARQUIVO .TXT ---
# O modo 'r' abre o arquivo para leitura
with open("relatorio.txt", "r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()
    print("\n--- Conteúdo Lido do Arquivo ---")
    print(conteudo)
```

### 2. Organizando Dados em Tabelas: Arquivos CSV (`.csv`)

Quando precisamos salvar dados organizados em linhas e colunas (como planilhas do Excel ou cadastros de clientes), o arquivo `.txt` simples não é suficiente. Usamos o formato **CSV** (*Comma-Separated Values* ou Valores Separados por Vírgula).

Para lidar com eles com segurança, o Python possui um módulo interno chamado `csv`.

```python
import csv

# Dados que queremos salvar em formato de tabela
dados_tabela = [
    ["Nome", "Cargo", "Idade"],
    ["Carlos Silva", "Desenvolvedor", 28],
    ["Ana Souza", "Engenheira de Dados", 32]
]

# --- ESCREVENDO EM UM CSV ---
with open("funcionarios.csv", "w", newline="", encoding="utf-8") as arquivo_csv:
    # Criamos o escritor de CSV
    escritor = csv.writer(arquivo_csv)
    
    # Escrevemos todas as linhas de uma vez só
    escritor.writerows(dados_tabela)

print("Arquivo 'funcionarios.csv' gerado com sucesso!")


# --- LENDO DE UM CSV ---
print("\n--- Lendo o Arquivo CSV ---")
with open("funcionarios.csv", "r", encoding="utf-8") as arquivo_csv:
    leitor = csv.reader(arquivo_csv)
    
    for linha in leitor:
        print("Linha lida:", linha)
```

### 3. O Padrão da Internet e das APIs: Arquivos JSON (`.json`)

Se os dicionários em Python tivessem uma versão física para salvar no computador, esse seria o **JSON** (*JavaScript Object Notation*). 

O JSON é o formato universal usado por praticamente todas as APIs modernas, webhooks e ferramentas de inteligência artificial para trocar dados estruturados. Ele preserva a estrutura de chaves e valores que aprendemos nos dicionários.

Para usá-lo, importamos a biblioteca `json` do Python:
* `json.dump()`: Salva um dicionário/lista do Python diretamente em um arquivo `.json`.
* `json.load()`: Lê um arquivo `.json` de volta para dentro do Python transformando em dicionário/lista.

```python
import json

# Um dicionário complexo (muito parecido com o que usamos em cadastros e APIs)
perfil_sistema = {
    "servidor": "localhost",
    "porta": 8080,
    "ativo": True,
    "administradores": ["Beatriz", "Carlos"]
}

# --- SALVANDO UM DICIONÁRIO EM JSON ---
# 'indent=4' serve apenas para deixar o arquivo bonito e legível para humanos
with open("configuracao.json", "w", encoding="utf-8") as arquivo_json:
    json.dump(perfil_sistema, arquivo_json, indent=4, ensure_ascii=False)

print("Arquivo 'configuracao.json' salvo com sucesso!")


# --- LENDO O ARQUIVO JSON DE VOLTA PARA O PYTHON ---
print("\n--- Lendo o Arquivo JSON ---")
with open("configuracao.json", "r", encoding="utf-8") as arquivo_json:
    dados_carregados = json.load(arquivo_json)
    
    # O Python converteu o JSON de volta para um dicionário normal!
    print("Tipo do dado carregado:", type(dados_carregados))
    print("Servidor configurado:", dados_carregados["servidor"])
    print("Lista de administradores:", dados_carregados["administradores"])
```
