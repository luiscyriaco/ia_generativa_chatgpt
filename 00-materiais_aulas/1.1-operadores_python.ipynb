## Operadores em Python 🔣

Os operadores são símbolos especiais utilizados para realizar operações em variáveis e valores, combinando números (`🔢`) e dados textuais (`🔤`) para realizar cálculos, comparações e atribuições.

## 1 Operadores de Atribuição

Os operadores de atribuição são usados para armazenar valores em variáveis, permitindo também atualizar o valor atual combinando uma operação matemática de forma mais compacta.

### Exemplos práticos:

```python
# 1. Atribuição simples (=)
x = 10
print("x inicial:", x)

# 2. Adição e atribuição (+=)
x = 2
x += 5  # O mesmo que: x = x + 5
print("Após += 5:", x)  # 7

# 3. Subtração e atribuição (-=)
x = 5
x -= 3  # O mesmo que: x = x - 3
print("Após -= 3:", x)  # 2

# 4. Multiplicação e atribuição (*=)
x = 5
x *= 2  # O mesmo que: x = x * 2
print("Após *= 2:", x)  # 10

# 4. Divisão e atribuição (/=)
x = 5
x /= 2  # O mesmo que: x = x / 2
print("Após *= 2:", x)  # 2.5

## 2 Operadores Aritméticos

Os operadores aritméticos são os símbolos que usamos no Python para realizar cálculos matemáticos básicos no dia a dia, como somar valores, calcular subtrações, multiplicar números, tirar a média com divisões e até calcular potências de forma rápida e direta.

| Operador | Operação |
| :---: | :--- |
| `+` | SOMA |
| `-` | SUBTRAÇÃO |
| `*` | MULTIPLICAÇÃO |
| `/` | DIVISÃO |
| `//` | DIVISÃO INTEIRA |
| `**` | EXPONENCIAÇÃO |
| `%` | MÓDULO |

````python
# Operações simples
a = 10
b = 3

print("Soma:", a + b)
print("Subtração:", a - b)
print("Multiplicação:", a * b)
print("Divisão:", a / b)
print("Divisão Inteira:", a // b)
print("Resto da divisão (Módulo):", a % b)
print("Potência:", a ** b) # 10 elevado a 3
print("Raiz quadrada de 10:", a ** (1/2)) # Raiz quadrada de 10

# Exemplos Módulo
modulo = 30 % 4
print("Resto da divisão de 30 por 4:", modulo)

modulo_negativo = -30 % 4
print("Resto da divisão de -30 por 4:", modulo_negativo)
````

### 2.1 Concatenação

A concatenação é o processo de juntar (unir) duas ou mais strings (textos) em uma única string. Em Python, fazemos isso utilizando o operador de adição (+).

````python
# 1. Concatenação simples de variáveis de texto
nome = "Maria"
sobrenome = "Silva"
nome_completo = nome + " " + sobrenome  # O " " serve para adicionar o espaço entre os nomes

print("Nome completo:", nome_completo)


# 2. Concatenação combinando texto com números (convertendo para string)
idade = 25
mensagem = "Olá, meu nome é " + nome + " e eu tenho " + str(idade) + " anos."

print(mensagem)

# 3. Números inseridos pelo usuário
nota_1 = input("Digite a primeira nota: ")
nota_2 = input("Digite a segunda nota: ")

media = (nota_1 + nota_2) / 2

print(f"A média das notas é: {media}")
````

## 3. Operadores Relacionais (Comparações)
Usamos operadores relacionais para comparar valores. O resultado de uma comparação é sempre um valor booleano: `True` (Verdadeiro) ou `False` (Falso).

| Operador | Operação |
| :---: | :--- |
| `>` | Maior que |
| `<` | Menor que |
| `>=` | Maior ou igual a |
| `<=` | Menor ou igual a |
| `==` | Igual a |
| `!=` | Diferente de |

````python
x = 10
y = 20

print("x é maior que y?", x > y)
print("x é menor ou igual a y?", x <= y)
print("x é igual a y?", x == y)
print("x é diferente de y?", x != y)
````

## 4. Operadores Lógicos
Podemos combinar várias condições usando operadores lógicos:

* `and` (E): Retorna `True` se **todas** as condições forem verdadeiras.

| Entrada A | Entrada B | Saída (AND) |
| :---: | :---: | :---: |
| F | F | F |
| F | V | F |
| V | F | F |
| V | V | V |

* `or` (OU): Retorna `True` se **pelo menos uma** das condições for verdadeira.

| Entrada A | Entrada B | Saída (OR) |
| :---: | :---: | :---: |
| F | F | F |
| F | V | V |
| V | F | V |
| V | V | V |

* `not` (NÃO): Inverte o valor (transforma `True` em `False` e vice-versa).

| Entrada A | Entrada B | Saída (NOT A) |
| :---: | :---: | :---: |
| F | F (negado) | V |
| V | F (negado) | F |

````python
# 1. Operador AND (E) - Retorna True apenas se AMBAS as condições forem verdadeiras
tem_sol = True
tem_dinheiro = True
vai_praia = tem_sol and tem_dinheiro
print("Vai à praia (AND):", vai_praia)  # True

# 2. Operador OR (OU) - Retorna True se PELO MENOS UMA condição for verdadeira
tem_carro = False
tem_bicicleta = True
pode_viajar = tem_carro or tem_bicicleta
print("Pode viajar (OR):", pode_viajar)  # True

# 3. Operador NOT (NÃO) - Inverte o valor lógico (True vira False, e vice-versa)
chovendo = False
fazer_caminhada = not chovendo
print("Fazer caminhada (NOT):", fazer_caminhada)  # True
````


## 5. Tomada de Decisão (Desvios Condicionais)

Agora que sabemos comparar valores, podemos usar o `if` para criar regras no nosso programa. O `elif` serve para testar uma nova condição caso a anterior seja falsa, e o `else` pega o que sobra última opção

### Exemplos:

Ir ao mercado: SE estiver chovendo, levo um guarda-chuva; SENÃO (se não estiver chovendo), vou de óculos de sol.<br>

Despertador: SE o despertador tocar, levanto da cama; SENÃO (se ele não tocar ou eu continuar dormindo), continuo na cama.<br>

Entrada no cinema: SE você tiver a idade mínima exigida pelo filme, você pode entrar; SENÃO (se for menor de idade), a entrada é proibida.<br>

````python
# 1. Exemplo simples com if e else
idade = 18

if idade >= 18:
    print("Você é maior de idade. Pode tirar a CNH!")
else:
    print("Você é menor de idade.")


# 2. Exemplo com if, elif e else (múltiplas condições)
nota = 7.5

if nota >= 7.0:
    print("Aluno aprovado!")
elif nota >= 5.0:
    print("Aluno em recuperação.")
else:
    print("Aluno reprovado.")


# 3. Exemplo de encadeamento / aninhamento (if dentro de outro if)
tem_cartao = True
saldo = 150.0

if tem_cartao:
    print("Cartão reconhecido.")
    
    if saldo >= 100.0:
        print("Compra aprovada! Saldo suficiente.")
    else:
        print("Compra negada. Saldo insuficiente.")
else:
    print("Erro: Nenhum cartão inserido.")
````
