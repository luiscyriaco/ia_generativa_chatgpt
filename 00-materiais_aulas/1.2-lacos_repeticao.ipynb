## Laços de Repetição (Loops) 🔄 

Os laços de repetição são estruturas usadas para executar um bloco de código várias vezes de forma automática, evitando que você precise repetir linhas idênticas manualmente.

### 1 O Laço FOR e a função range()
O for em Python é muito utilizado junto com a função range() para repetir uma ação um número específico de vezes. O range() nos permite controlar onde a contagem começa, onde termina e qual é o intervalo (passo) entre os números.

Podemos criar 3 tipos de montagens para iteração através da função range()

1. Com 1 único argumento que permite indicar apenas a quantidade de loopings que o for realizará, partindo de 0.
```Python
range(qtd_loopings)
range(3)
```

2. Com 2 argumentos que permite indicar 1-o valor inicial de **i** e 2-o valor final de **i**.
Valendo ressaltar que após as iterações quando o valor de **i** alcança o mesmo valor final escolhido ele para e não acontece a última iteração.
Dessa forma o valor final deverá sempre ser maior do que a quantidade de iterações desejadas.
```Python
range(início,fim)
# Nesse exemplo a iteração ocorrerá somente até a nona iteração
range(1,10)
# Se o desejo for de 10 iterações, deverá definir 11 para o valor final
range(1,11)
```

3. Com 3 argumentos que permite além de determinar o início e o fim das iterações, determinar o tamanho do salto, como por exemplo saltar de 2 e 2.
```Python
# Nesse exemplo os valores de i irá salta de 2 em 2, recebendo apenas valores pares
range(início,fim,salto)
range(2,11,2)
```

````python
# 1. range(3): Começa no 0 por padrão e vai até 2 (total de 3 repetições)
print("Exemplo 1 - range(3):")
for i in range(3):
    print("Valor de i:", i)


# 2. range(1, 5): Define o início (1) e o fim (para em 4, pois o 5 não é incluído)
print("\nExemplo 2 - range(1, 5):")
for i in range(1, 5):
    print("Valor de i:", i)


# 3. range(1, 10, 2): Define o início (1), o fim (menos que 10) e o salto/passo (pula de 2 em 2)
print("\nExemplo 3 - range(1, 10, 2):")
for i in range(1, 10, 2):
    print("Valor de i:", i)
````

### 1.1 Percorrendo Elementos com o FOR 📄

Além de contar números, o `for` é extremamente útil para percorrer elementos de uma lista, como páginas de um site ou linhas de um texto.

### 2 O Laço WHILE 🔄

O while (enquanto) é um laço de repetição ideal para quando não sabemos exatamente quantas vezes o código precisará repetir, mas sabemos a condição que precisa ser verdadeira para ele continuar rodando.

Enquanto a condição for verdadeira (True), o bloco de código dentro do while é executado repetidamente. É fundamental atualizar a variável de controle dentro do loop para evitar um loop infinito (quando o programa trava rodando para sempre).

````python
# 1. Exemplo básico: Contagem controlada por uma variável
contador = 1

while contador <= 3:
    print("Contador atual:", contador)
    contador += 1  # Incrementa para a condição deixar de ser verdadeira em algum momento

print("Fim do loop while básico!\n")

# 2. Exemplo intermediário: Solicitando repetição baseada em uma escolha do usuário
resposta = "sim"

while resposta.lower() == "sim":
    print("Executando a tarefa...")
    # Simulando a pergunta ao usuário se deseja continuar
    resposta = input("Deseja continuar? (sim/nao): ")

print("Programa encerrado!")


# 3. Exemplo simulando validação de dados (repetindo até acertar)
senha_correta = "python123"
tentativa = ""

# Simulando tentativas para o exemplo
tentativas_feitas = ["1234", "senha_errada", "python123"]
indice = 0

while tentativa != senha_correta:
    tentativa = tentativas_feitas[indice]
    print(f"Tentativa com '{tentativa}': Acesso negado.")
    indice += 1
    if tentativa == senha_correta:
        print("Acesso concedido! 🎉")
        break
````

#### 2.1. Validação de Senha em Sistema de Login (O clássico "Até acertar")

````python
senha_cadastrada = "1234"
tentativas = 3

while tentativas > 0:
    senha_digitada = input("Digite sua senha: ")
    
    if senha_digitada == senha_cadastrada:
        print("Acesso permitido! Bem-vindo. 🎉")
        break
    else:
        tentativas -= 1
        print(f"Senha incorreta. Você ainda tem {tentativas} tentativa(s).")

if tentativas == 0:
    print("Acesso bloqueado por excesso de tentativas! 🚨")
````

#### 2.2. Tabuada Automática (Matemática com um único bloco)

````python
numero = int(input("Digite um número para ver a tabuada: "))

print(f"\n--- Tabuada do {numero} ---")
for i in range(1, 11):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
````

#### 2.3. Criando Padrões Visuais / Desenhos no Terminal (Arte ASCII)

````python
print("Desenhando uma escada:")
for i in range(1, 6):
    print("*" * i)
````

#### 2.4 Importando tkinter

Para abrir uma janela gráfica real de verdade, nós usamos o while integrado com uma biblioteca. Veja como fica um exemplo real usando o Tkinter (que já vem instalado por padrão no Python):

````python
import tkinter as tk

# Cria a janela principal
janela = tk.Tk()
janela.title("Minha Janela com While/Loop")
janela.geometry("300x200")

# O Tkinter possui seu próprio loop interno embutido (mainloop)
# que funciona exatamente com a mesma lógica de manter a janela rodando:
print("A janela foi aberta. Feche a janela para continuar o programa.")
janela.mainloop()

print("A janela foi fechada pelo usuário!")
````

#### 2.5 Importando Pygame

E se estivéssemos usando o Pygame para jogos (que você também costuma explorar), o código com o while que faz a janela rodar de verdade seria assim:

````python
import pygame

pygame.init()
tela = pygame.display.set_mode((400, 300))
pygame.display.set_caption("Janela com Pygame")

rodando = True

# O while aqui é o coração do jogo: roda o loop de quadros (frames) sem parar
while rodando:
    # Monitora os eventos (como clicar no 'X' para fechar)
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            rodando = False

    # Atualiza a tela do jogo
    pygame.display.flip()

pygame.quit()
````
