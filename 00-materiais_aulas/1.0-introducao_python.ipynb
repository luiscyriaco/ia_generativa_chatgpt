````python
# python
numero = 5
resultado = numero * 2

print("O resultado é:", resultado)
````

### 1.2 Linguagem de Baixo Nível (Assembly)
Em Assembly, não existe um comando simples como print. Para mostrar um número na tela, o programa de baixo nível precisa primeiro converter o número em caractere de texto.

````python
"""
section .data
    msg db "O resultado e: 10", 0Ah  ; Precisamos escrever a frase pronta na memória

section .text
    global _start

_start:
    ; 1. A conta em si (multiplicar 5 por 2)
    mov al, 5
    shl al, 1          ; Al agora vale 10 (mas ainda é um dado binário no processador)

    ; 2. Para imprimir, o SO exige chamar o kernel (syscall de escrita)
    mov rax, 1         ; Código 1 = sys_write (escrever na tela)
    mov rdi, 1         ; Descritor 1 = Saída padrão (stdout / monitor)
    mov rsi, msg       ; Endereço da mensagem na memória
    mov rdx, 18        ; Tamanho da mensagem em bytes
    syscall            ; Dispara a ordem direta para o Kernel do Sistema Operacional

    ; 3. Encerrar o programa
    mov rax, 60        ; Código 60 = sys_exit (sair)
    xor rdi, rdi       ; Código de saída 0 (sem erro)
    syscall
"""
````

### 1.3 Diferença entre Linguagens Compiladas e Interpretadas

* **Linguagem Compilada** (o livro traduzido antes): Antes de você rodar o programa, um programa chamado **compilador** traduz o seu código inteiro de uma vez só para um arquivo executável de máquina (como um `.exe`). Só depois disso você consegue rodar.

* **Linguagem Interpretada** (o intérprete de conferência): Você aperta o *play* e o **interpretador** vai traduzindo e executando o seu comando na hora, na frente do processador, sem precisar criar um arquivo executável separado antes.

````python
print("Olá, Mundo! Bem-vindo ao Python.")
````

## 2 Variáveis e Tipos de Dados

### 2.1 Variáveis
As **variáveis** servem para armazenar informações na memória. Em Python, não precisamos declarar o tipo da variável explicitamente.

nome_aluno   |   snake_case<br>
nomeAluno    |   camelCase


````python
# Criando variáveis de diferentes tipos
nome = "Ana"        # String (texto)
idade = 25          # Inteiro (número inteiro)
altura = 1.75       # Float (número decimal)
estudando = True    # Booleano (Verdadeiro ou Falso)

print("Nome:", nome, "| Tipo:", type(nome))
print("Idade:", idade, "| Tipo:", type(idade))
print("Altura:", altura, "| Tipo:", type(altura))
print("Estudando:", estudando, "| Tipo:", type(estudando))
````

## 3 Coleções de Dados

As coleções de dados são estruturas que permitem armazenar e organizar múltiplos valores juntos, facilitando a manipulação de grandes volumes de informações em vez de usar variáveis separadas.

### 3.1 Lista, tupla e dicionário

````python
# 1. Lista (modificável)
frutas = ["maçã", "banana", "laranja"]

# 2. Tupla (fixa/imutável)
coordenadas = (10.5, 20.3)

# 3. Dicionário (chave: valor)
pessoa = {"nome": "Carlos", "idade": 30}
````

#### 3.1.1. Acesso a Listas
As listas usam índices numéricos que começam em 0. Você pode acessar os elementos informando a posição entre colchetes.

````python
frutas = ["maçã", "banana", "laranja", "uva"]

print(frutas[0])   # Primeiro elemento: maçã
print(frutas[2])   # Terceiro elemento: laranja
print(frutas[-1])  # Último elemento: uva (índice negativo conta de trás para frente)
````

#### 3.1.2. Acesso a Tuplas
Assim como as listas, as tuplas usam índices numéricos a partir de 0 para recuperar os valores.

````python
coordenadas = (10.5, 20.3, 40.1)

print(coordenadas[0])  # Primeiro valor: 10.5
print(coordenadas[1])  # Segundo valor: 20.3
````

#### 3.1.3. Acesso a Dicionários
Os dicionários não usam posições numéricas, mas sim as chaves que você criou para cada dado.

**Método .get()**
Ao tentarmos acessar uma determinada chave: inexistente em um dicionário, iremos nos deparar com um erro "KeyError".<br>
Para evidar esse problema, podemos usar a função .get(), e caso a chave seja inexistente a função .get() nos retornará o valor "none".<br>
Podemos dentro da função .get() informar uma mensagem de erro, caso a chave seja inexistênte. Como no exemplo a seguir:

```Python
dicionario.get("chave")  # Retorna None em vez de erro
dicionario.get("chave", "valor_padrao")  # Retorna "valor_padrao" se não existir

````python
pessoa = {"nome": "Carlos", "idade": 30, "cidade": "São Paulo"}

print(pessoa["nome"])      # Acessando pela chave: Carlos
print(pessoa.get("idade"))  # Usando o método .get(): 30
````

### 3.2 Matrizes (Listas de Listas)

As matrizes são estruturas bidimensionais com linhas e colunas, permitindo organizar dados em formato de tabela usando listas dentro de listas.
Podemos comparar a estrutura de uma matriz ao padrão utilizado no Excel.

            A       B      C
| 1 | descrição | preço | qtd. |
| :---: | :--- | ---: | ---: |
| 2 | Mouse | 50.0 | 15 |
| 3 | Teclado | 120.0 | 8 |
| 4 | Monitor | 900.0 | 4 |



Quando iteramos por uma matriz, iniciamos a contagem apartir do valor zero tanto no sentido da linha, quanto no sentido da coluna.

````python
# Exemplo 1: Tabela de produtos (Linha = Produto, Colunas = [Nome, Preço, Estoque])
produtos = [
    #   0       1   2
"0"  ["Mouse", 50.0, 15],
"1"  ["Teclado", 120.0, 8],
"2"  ["Monitor", 900.0, 4]
]
    #                               L  C
print("Preço do teclado:", produtos[1][1])  # Acessa a linha 1, coluna 1

# L = linha
# C = coluna
````

````python
# Exemplo 2: Jogo da Velha (Tabuleiro 3x3)
tabuleiro = [
    ["X", "O", " "],
    [" ", "X", "O"],
    ["O", " ", "X"]
]
print("Centro do tabuleiro:", tabuleiro[1][1])  # Acessa a linha 1, coluna 1
````

````python
# Exemplo 3: Notas de alunos (Linha = Aluno, Colunas = [Nome, Nota 1, Nota 2])
notas_turma = [
    ["Ana", 8.5, 9.0],
    ["Bruno", 6.0, 7.5],
    ["Carla", 9.5, 10.0]
]
print("Segunda nota da Carla:", notas_turma[2][2])  # Acessa a linha 2, coluna 2
````

### 3.3 Dicionário Composto

Em um dicionário composto, podemos observar o aninhamento de dicionários em Python. Mas a estrutura de "chaves":"valores" permanece a mesma de um dicionário comum.<br>
No exemplo a seguir temos a chave 101: que abriga dentro de si o valor de outro dicionário.<br>
Podemos observar inclusive que detro da chave: habilidades: podemos criar uma lista de valores.<br>
Essa estrutura complexa, segue o mesmo padrão de construção de um arquivo JSON.

````python
funcionarios = {
    101: {
        "nome": "Carlos",
        "cargo": "desenvolvedor",
        "habilidades": ["Python","C##","Java"]
    },
    102: {
        "nome": "Mariana",
        "cargo": "gerente de projetos",
        "habilidades": ["Scrum","Gestão"]
    }
}

print(funcionarios[101]["cargo"])
print(funcionarios.get(103,{}).get("nome","Fucionário não encontrado"))
````
